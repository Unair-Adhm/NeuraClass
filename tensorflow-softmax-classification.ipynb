```python
import tensorflow as tf

vector1 = tf.random.uniform(shape=(10,), minval=0, maxval=1)

print("Vector 1:\n", vector1)
```

```python
mean = tf.reduce_mean(vector1)
std_dev = tf.math.reduce_std(vector1)
max_val = tf.reduce_max(vector1)

print("Mean:", mean)
print("Standard Deviation:", std_dev)
print("Maximum Value:", max_val)
```

```python
matrix = tf.random.uniform(shape=(5, 5), minval=0, maxval=1)

transpose_matrix = tf.transpose(matrix)

print("Matrix:\n", matrix)
print("Transpose of Matrix:\n", transpose_matrix)
```

```python
matrix_mult = tf.matmul(matrix, matrix)

print("Matrix multiplied by itself:\n", matrix_mult)
```

```python
mean_matrix = tf.reduce_mean(matrix)
std_matrix = tf.math.reduce_std(matrix)

normalized_matrix = (matrix - mean_matrix) / std_matrix

print("Normalized Matrix:\n", normalized_matrix)
```

```python
# Step 1: Created a vector of 10 random numbers using TensorFlow's tf.random.uniform function.
# Step 2: Calculated the mean, standard deviation, and maximum value of the vector using tf.reduce_mean, tf.math.reduce_std, and tf.reduce_max.
# Step 3: Generated a 5x5 random matrix with tf.random.uniform and computed its transpose with tf.transpose.
# Step 4: Multiplied the matrix by itself using tf.matmul to perform proper matrix multiplication.
# Step 5: Normalized the matrix using the formula (X - mean) / standard deviation, so that the resulting matrix has a mean approximately 0 and standard deviation approximately 1.
```

```python
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.utils import to_categorical

(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train / 255.0
x_test = x_test / 255.0

y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

print("Training samples shape:", x_train.shape)
print("Training labels shape:", y_train.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense

model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

model.summary()
```

```python
history = model.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.1)
```

```python
import matplotlib.pyplot as plt

plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Accuracy over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Loss over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.show()
```
